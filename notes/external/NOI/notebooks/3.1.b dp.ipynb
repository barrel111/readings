### Memoizing Fibonacci.

The problem with the recursive implementation of Fibonacci was that we would redundantly recompute known values. Memoization (derived from the word memo, not a play on memorization) fixes this flaw by explicitly storing computed values in a table and performing a look up in the table before making recursive calls.

```python
fib_tbl = [0, 1]
def fib_memo(n):
    if len(fib_tbl) <= n:
        fib_tbl.extend([None] * (n - len(fib_tbl) + 1))
   
    if fib_tbl[n] is not None:
       return fib_tbl[n]
    fib_tbl[n] = fib_memo(n-1) + fib_memo(n-2)
   
    return fib_tbl[n]
```

```python
fib_memo(100)
```

### DP I: Weighted Independent Set on a Path.

Memoization is a general caching technique that can speed up any recursive function by storing results. ### DP II: Weighted Interval Scheduling.

We now consider a more involved dynamic program. 

**Problem.** Given a set of jobs $\mathcal{J}$ where each job $j \in \mathcal{J}$ is specified by a starting time $s_j$, finishing time $f_j$ and a value $v_j$. Two jobs are compatible if they do not overlap. Find a maximum weight subset of compatible jobs.

**How would you approach this problem?**


Here is a helpful observation. First, suppose the jobs are labelled $j_1, \dots, j_k$ in nondecreasing finish time i.e. $f_{j_1} \le f_{j_2} \le \dots \le f_{j_k}$. Let $b(i)$ be the largest index $\ell < i$ such that $j_\ell$ and $j_i$ are compatible. That is, if we schedule a job $j_i$, then $j_{b_i}$ is the latest finishing job that we can schedule prior to $j_i$. If $\mathcal{O}$ is a maximum weight subset of compatible jobs, we have the following observation: either job $j_k \in \mathcal{O}$ or $j_k \notin \mathcal{O}$. Particularly, if $j_k \in \mathcal{O}$ the remaining jobs can only be scheduled from $j_1, \dots, j_{b(k)}$. Otherwise, if $j_k \notin \mathcal{O}$ the remaining jobs can only be scheduled from $j_1, \dots, j_{k - 1}$. If we knew how to solve the subproblems corresponding to these subsets of the job, we would be able to determine whether $j_k$ is in $\mathcal{O}$ by comparing the weight of the corresponding solutions.

This provides us with a recursive formulation for the solution. Let $\tt{OPT}(i)$ denote the weight of the maximum weight compatible subset of the jobs $j_1, \dots, j_i$. Then, from the prior discussion we have, 
\begin{align*}
\tt{OPT}(i) = \max \begin{cases}
s_i + \tt{OPT}(b(i)), \\ \tt{OPT}(i - 1).
\end{cases}
\end{align*}

Furthermore, if we only have no jobs, the optimal subset is the empty set itself of weight $0$. Thus, we have the base case $\tt{OPT}(0) = 0$.
